```python
from sklearn.preprocessing import LabelEncoder

data = ['Red', 'Green', 'Blue', 'Red','purple','pink']
le = LabelEncoder()
encoded = le.fit_transform(data)
print(encoded)
```

```python

from sklearn.preprocessing import OneHotEncoder
import numpy as np

data = np.array([['Red'],['pink'], ['Green'], ['Blue'], ['Red']])
ohe = OneHotEncoder()
encoded = ohe.fit_transform(data)
print(encoded)
```
